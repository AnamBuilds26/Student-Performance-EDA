```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/StudentsPerformance.csv')

print("Dataset loaded successfully!")
print("Number of rows and columns:", df.shape)

df.head()
```

```python

# Dataset information
print("Dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
display(df.describe())
```

```python

plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="math score", bins=20, kde=True)

plt.title("Distribution of Students' Math Scores")
plt.xlabel("Math Score")
plt.ylabel("Number of Students")

plt.show()
```

Interpretation: The histogram shows the distribution of students' math scores. It helps us identify the most common score ranges, the spread of scores, and whether the distribution is approximately symmetric or skewed.

```python

plt.figure(figsize=(9, 5))

sns.histplot(df["math score"], color="skyblue",
             label="Math", kde=True, element="step")
sns.histplot(df["reading score"], color="orange",
             label="Reading", kde=True, element="step")
sns.histplot(df["writing score"], color="green",
             label="Writing", kde=True, element="step")

plt.title("Distribution of Math, Reading and Writing Scores")
plt.xlabel("Score")
plt.ylabel("Number of Students")
plt.legend()
plt.show()
```

Interpretation: This graph compares the distributions of math, reading, and writing scores. It helps identify differences in score ranges and where student scores are concentrated.

```python

fig, axes = plt.subplots(2, 2, figsize=(12, 9))

sns.countplot(data=df, x="gender", ax=axes[0, 0])
axes[0, 0].set_title("Students by Gender")

sns.countplot(data=df, x="race/ethnicity", ax=axes[0, 1])
axes[0, 1].set_title("Students by Race/Ethnicity")

sns.countplot(data=df, x="lunch", ax=axes[1, 0])
axes[1, 0].set_title("Students by Lunch Type")

sns.countplot(data=df, x="test preparation course",
              ax=axes[1, 1])
axes[1, 1].set_title("Test Preparation Course")

plt.tight_layout()
plt.show()
```

Interpretation: These bar charts compare the number of students in each category. They help identify the most common gender, race/ethnicity group, lunch type, and test preparation status. Differences in category sizes are important when interpreting score comparisons.

```python

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

sns.boxplot(data=df, x="gender", y="math score", ax=axes[0])
axes[0].set_title("Math Scores by Gender")

sns.boxplot(data=df, x="gender", y="reading score", ax=axes[1])
axes[1].set_title("Reading Scores by Gender")

sns.boxplot(data=df, x="gender", y="writing score", ax=axes[2])
axes[2].set_title("Writing Scores by Gender")

plt.tight_layout()
plt.show()
```

Interpretation: The boxplots compare the median scores, spread, and potential outliers for male and female students. They help identify differences in performance across subjects. The results describe this dataset and should not be generalized to every student.

```python

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

sns.boxplot(data=df, x="test preparation course",
            y="math score", ax=axes[0])
axes[0].set_title("Math Score vs Test Preparation")

sns.boxplot(data=df, x="test preparation course",
            y="reading score", ax=axes[1])
axes[1].set_title("Reading Score vs Test Preparation")

sns.boxplot(data=df, x="test preparation course",
            y="writing score", ax=axes[2])
axes[2].set_title("Writing Score vs Test Preparation")

plt.tight_layout()
plt.show()
```

Interpretation: These boxplots compare students who completed the test preparation course with those who did not. Differences in medians and score distributions can suggest an association between preparation and performance, but do not prove that the course caused the differences.

```python

# Select numerical columns
numeric_df = df.select_dtypes(include="number")

# Calculate correlations
correlation = numeric_df.corr()

# Draw heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    fmt=".2f"
)

plt.title("Correlation Heatmap of Student Scores")
plt.show()
```

**Interpretation of Correlation Heatmap**

The correlation heatmap shows the strength and direction of relationships between numerical variables. A value near +1 indicates a strong positive relationship, a value near -1 indicates a strong negative relationship, and a value near 0 indicates a weak linear relationship.

The relationships between math, reading, and writing scores are particularly important. The heatmap helps identify whether students who perform well in one subject also tend to perform well in other subjects. Correlation indicates association, not causation.

```python

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="math score",
    y="reading score",
    hue="gender",
    alpha=0.7
)

plt.title("Relationship Between Math and Reading Scores")
plt.xlabel("Math Score")
plt.ylabel("Reading Score")
plt.show()
```

**Interpretation:** The scatter plot shows the relationship between math and reading scores. An upward pattern indicates a positive relationship, meaning students with higher math scores tend to have higher reading scores. The colors allow us to compare the patterns for male and female students.

```python

sns.pairplot(
    df,
    vars=["math score", "reading score", "writing score"],
    hue="test preparation course",
    diag_kind="hist"
)

plt.show()
```

**Interpretation:** The pair plot compares all three subject scores together. The scatter plots show relationships between pairs of subjects, while the diagonal histograms show individual score distributions. The colors compare students who completed the test preparation course with those who did not.

```python

plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df[["math score", "reading score", "writing score"]]
)

plt.title("Outlier Detection in Student Scores")
plt.ylabel("Score")
plt.show()

# Count potential outliers using the IQR method
for col in ["math score", "reading score", "writing score"]:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(col, "- potential outliers:", len(outliers))
```

**Interpretation:** Boxplots help identify unusually low or high scores. The IQR method flags observations outside the expected range based on the middle 50% of scores. These observations are potential outliers, not necessarily errors, and should not be removed without investigation.
